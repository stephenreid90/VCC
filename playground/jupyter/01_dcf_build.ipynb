```python
import sys
sys.path.insert(0, '../..')
import pandas as pd
from models.dcf import DCFModel, DCFAssumptions
print('Libraries loaded ✅')
```

```python
# === EDIT THESE ASSUMPTIONS ===
assumptions = DCFAssumptions(
    fcf_base=110e9,        # Base year FCF in USD
    fcf_growth_rate=5.0,   # % per year
    wacc=7.65,             # %
    terminal_growth=2.5,   # %
    explicit_periods=5,    # Years
    net_debt=-50e9,        # Negative = net cash
    shares_outstanding=15.5e9,  # Absolute share count
)
```

```python
model = DCFModel(assumptions)
result = model.calculate()

print(f'Enterprise Value:  ${result.enterprise_value/1e9:.1f}B')
print(f'Equity Value:      ${result.equity_value/1e9:.1f}B')
print(f'Price per Share:   ${result.price_per_share:.2f}')
print(f'PV Explicit:       ${result.pv_explicit/1e9:.1f}B')
print(f'PV Terminal:       ${result.pv_terminal/1e9:.1f}B')
```

```python
# FCF projections table
df = pd.DataFrame({
    'Year': range(1, len(result.fcf_projections)+1),
    'FCF (USD bn)': [f/1e9 for f in result.fcf_projections]
})
df.style.format({'FCF (USD bn)': '{:.1f}'})
```

```python
# Sensitivity analysis
sens = result.sensitivity_wacc_growth
df_sens = pd.DataFrame(list(sens.items()), columns=['Scenario', 'Enterprise Value (USD bn)'])
df_sens['Enterprise Value (USD bn)'] = df_sens['Enterprise Value (USD bn)'] / 1e9
df_sens
```
